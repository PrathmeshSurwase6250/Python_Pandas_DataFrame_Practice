Create a cleaning function that reports missing values before and after cleaning.

```python

def drop(df):
    print("Missing values before cleaning:")
    print(df.isnull().sum())

    df_cleared = df.dropna().copy()

    print("\nMissing values after cleaning:")
    print(df_cleared.isnull().sum())

    return df_cleared

students_cleaned = drop(students)
```
