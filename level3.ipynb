```python
import pandas as pd
```

```python
students = pd.read_csv('Pandas_Practice_Pack/students.csv')
```

```python
students.head(2)
```

Select the first 5 rows using iloc.

```python
students.iloc[:5]
```

Select rows at positions 5 through 10.

```python
students.iloc[4:15]
```

Select the first 5 rows and the first 3 columns.

```python
students.iloc[:5,0:3]
```

Find students whose marks are greater than 70.

```python
students[students['marks'] > 70]
```

Find students whose marks are less than 50.

```python
students[students['marks'] <= 50]
```

Find students whose marks equal 80.

```python
students[students['marks'] == 80]
```

Find students from Pune.

```python
students[students['city'] == 'Pune']
```

Find students from Pune who scored more than 70.

```python
students[(students['city'] == 'Pune') & (students['marks'] >= 70)]
```

Find students with marks greater than 80 and attendance greater than 75.

```python
students[(students['marks'] > 80) & (students['attendance'] > 75)]
```

Find students from Pune or Mumbai.

```python
students[students['city'].isin(['Pune' , "Mumbai"])]
```

```python
students[(students['city'] == 'Pune') | (students['city'] == 'Mumbai')]
```

Find students studying Computer or IT.

```python
students[students['course'].isin(['IT' , 'Computer'])]
```

Sort students by marks in ascending order.

```python
students.sort_values(by='marks')
```

Sort students by marks in descending order.

```python
students.sort_values(by='marks' , ascending=False)
```

Find the 5 students with the highest marks.

```python
students.sort_values(by='marks' , ascending=False).head(5)
```

Find the 5 students with the lowest marks.

```python
students.sort_values(by='marks' , ascending=False).tail(5)
```

Sort by course and then marks in descending order.

```python
students.sort_values(by=['course' , 'marks'] , ascending=[True , False])
```

Find students aged 20 or above.

```python
students[students['age'] >= 20]
```

Find students whose marks are between 60 and 80, inclusive.

```python
students[students['marks'].apply(lambda x : x > 60 and  x < 80)]
```

Filter students using isin() for a selected list of cities.

```python
students[students['city'].isin(['Pune'])]
```

Select the student with a specific student_id using loc after setting that column as the index.

```python
students.loc[students['student_id'] == 'ST003']
```
