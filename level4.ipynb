Create a reusable function that classifies marks into performance categories.

```python
def classify_performance(marks):
    if marks >= 90:
        return 'Excellent'
    elif marks >= 80:
        return 'Very Good'
    elif marks >= 70:
        return 'Good'
    elif marks >= 60:
        return 'Average'
    else:
        return 'Poor'
```

```python
students['performance'] = students['marks'].apply(classify_performance)
```
