```python
import numpy as np
import pandas as pd 
```

```python
students = pd.read_csv('Pandas_Practice_Pack/students.csv')
students.head()
```

Write a function that returns "Pass" or "Fail" from a mark.

```python
def Check_pass_Fail(marks) :
    return 'Pass' if marks >= 40 else 'Fail'
```

```python
students['Status_Of_Result']= students['marks'].apply(Check_pass_Fail)
students
```

Use apply() to create a pass/fail column.

```python
students['By_Apply_CHECK'] = students['marks'].apply(lambda x : 'Pass' if  x >= 40 else 'Fail')
students
```

Use lambda to add 10 to each mark.

```python
students['marks'] = students['marks'].apply(lambda x : x + 10)
students
```

Use map() to shorten placement-status labels.

```python
status_map = {
    'Placed': 'Yes',
    'Not Placed': 'No'
}
students['placement_status'].map(status_map)
```

Use np.where() to create a binary pass/fail column.

```python
students['binary'] = np.where(students['marks'] > 40 , "pass" ,'Fail')
```

Use np.select() to classify students into multiple performance categories.

```python


conditions = [
    students['marks'] >= 90,
    students['marks'] >= 80,
    students['marks'] >= 70,
    students['marks'] >= 60
]

categories = [
    'Excellent',
    'Very Good',
    'Good',
    'Average'
]

students['performance'] = np.select(
    conditions,
    categories,
    default='Poor'
)
```

Write a function that accepts a DataFrame and column name and returns its mean.

```python
def mean(df ,column) :
    return df[column].mean()

mean(students , 'marks')
```

Write a function that returns the top N students by marks.  

```python
def top(df , n):
    return df[n].sort_values( ascending=False).head(5)
top(students , 'marks')
```
